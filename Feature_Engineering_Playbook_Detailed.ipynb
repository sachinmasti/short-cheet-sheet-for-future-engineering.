## 🔹 ENCODING TECHNIQUES


### 1️⃣ One-Hot Encoding (Nominal Data)

📌 What?
Converts each category into separate binary columns.

📌 Why?
Prevents model from assuming ordinal relationship.

📌 When to use?
- Nominal categorical data
- Small number of categories

📌 When NOT to use?
- High-cardinality features (1000+ categories)


```python

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

df = pd.DataFrame({'color': ['Red', 'Blue', 'Green', 'Blue']})

ohe = OneHotEncoder(drop='first', sparse=False)
encoded = ohe.fit_transform(df[['color']])

print("Encoded Data:")
print(encoded)
```


### 2️⃣ Ordinal Encoding

📌 What?
Assigns ordered numbers to categories.

📌 Why?
Preserves ranking relationship.

📌 When to use?
- Ordered categories like Low < Medium < High

📌 Warning:
Do NOT use for nominal categories.


```python

from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({'education': ['Low', 'Medium', 'High']})

oe = OrdinalEncoder(categories=[['Low','Medium','High']])
encoded = oe.fit_transform(df[['education']])

print(encoded)
```


### 3️⃣ Frequency Encoding

📌 What?
Replaces category with its frequency.

📌 Why?
Reduces dimensionality in high-cardinality features.

📌 When to use?
- Large datasets
- High unique categories (City, Pincode)

📌 Risk:
Does not preserve target relationship.


```python

df = pd.DataFrame({'city': ['Mumbai','Delhi','Mumbai','Pune']})

freq = df['city'].value_counts(normalize=True)
df['city_freq'] = df['city'].map(freq)

print(df)
```

## 🔹 TRANSFORMATIONS


### 4️⃣ Log Transformation

📌 What?
Applies log to compress large values.

📌 Why?
Reduces skewness and impact of outliers.

📌 When to use?
- Right-skewed data (Income, Population)

📌 Use log1p when zeros exist.


```python

import numpy as np

df = pd.DataFrame({'income': [1000, 5000, 10000, 50000]})
df['income_log'] = np.log1p(df['income'])

print(df)
```


### 5️⃣ Power Transformation (Yeo-Johnson)

📌 What?
Transforms data to be more Gaussian-like.

📌 Why?
Improves performance of linear models.

📌 When to use?
- Highly skewed data
- Linear regression assumptions required


```python

from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method='yeo-johnson')
X_transformed = pt.fit_transform(df[['income']])

print(X_transformed)
```


### 6️⃣ Polynomial Features

📌 What?
Creates higher-degree features (x², x³).

📌 Why?
Helps linear models capture nonlinear relationships.

📌 When to use?
- When relationship is curved


```python

from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform([[2],[3],[4]])

print(X_poly)
```

## 🔹 SCALING TECHNIQUES


### 7️⃣ Standard Scaling

📌 What?
Mean = 0, Std = 1

📌 Why?
Helps gradient-based models converge faster.

📌 When to use?
- Logistic Regression
- Neural Networks
- SVM


```python

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled = scaler.fit_transform([[10],[20],[30]])

print(scaled)
```


### 8️⃣ MinMax Scaling

📌 What?
Scales values between 0 and 1.

📌 When to use?
- Neural Networks
- Image data


```python

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaled = scaler.fit_transform([[10],[20],[30]])

print(scaled)
```


### 9️⃣ Robust Scaling

📌 What?
Uses median & IQR.

📌 When to use?
- When dataset has many outliers.


```python

from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()
scaled = scaler.fit_transform([[10],[20],[1000]])

print(scaled)
```


## 🔹 Cyclical Encoding

📌 What?
Uses sine & cosine to encode repeating cycles.

📌 When to use?
- Day of week
- Month
- Hour
- Angle data

📌 Why?
Preserves circular relationship.


```python

import numpy as np

#iss main agar monts ke liye nikalna hai to days ki jagah per month kar dene se hota hia.
days = np.array([0,1,2,3,4,5,6])
sin_day = np.sin(2 * np.pi * days / 7)
cos_day = np.cos(2 * np.pi * days / 7)

print("Sin:", sin_day)
print("Cos:", cos_day)
```


## 🔹 Feature Interaction & Ratios

📌 What?
Creates relationship features.

📌 Why?
Sometimes interaction is more powerful than single feature.

📌 Example:
Area per room is more informative than area alone.


```python

df = pd.DataFrame({
    'area':[1000,1500,2000],
    'rooms':[2,3,4]
})

df['area_per_room'] = df['area'] / df['rooms']
df['area_x_rooms'] = df['area'] * df['rooms']

print(df)
```


## 🎯 Interview Summary

- Always split data before fitting transformers.
- Avoid data leakage.
- One-Hot for nominal.
- Ordinal for ordered data.
- Frequency/Target for high-cardinality.
- Scaling required for gradient-based models.
- Tree models less sensitive to scaling.
- Feature engineering often more important than model selection.
